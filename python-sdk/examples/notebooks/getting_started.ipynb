# Commit-Relay SDK - Getting Started

This notebook demonstrates the basic usage of the commit-relay Python SDK.

## Installation

Make sure the SDK is installed:

```bash
cd python-sdk
pip install -e .
```

## 1. Initialize the Client

````python
from commit_relay import CommitRelayClient

# Initialize client (adjust URL if needed)
client = CommitRelayClient(base_url='http://localhost:3000')

# Test connectivity
if client.ping():
    print("✓ Connected to commit-relay dashboard")
else:
    print("✗ Cannot connect to dashboard")
````

## 2. Get Current Metrics

````python
# Fetch current metrics
metrics = client.metrics.get_current()

print("Current System Metrics:")
print(f"  Active Workers: {metrics.get('active_workers', 0)}")
print(f"  Tasks Pending: {metrics.get('tasks_pending', 0)}")
print(f"  Tasks In Progress: {metrics.get('tasks_in_progress', 0)}")
print(f"  Success Rate: {metrics.get('success_rate', 0):.1f}%")
````

## 3. Analyze Workers

````python
# Get worker statistics
stats = client.workers.get_stats()

print(f"Worker Pool Statistics:")
print(f"  Total: {stats['total']}")
print(f"  Active: {stats['active']}")
print(f"\nBy Type:")
for worker_type, count in stats['by_type'].items():
    print(f"  - {worker_type}: {count}")
````

## 4. Export to DataFrames

````python
import pandas as pd

# Export metrics to DataFrame
metrics_df = client.metrics.to_dataframe(hours=24)

print(f"Metrics DataFrame: {len(metrics_df)} rows")
print(f"Columns: {list(metrics_df.columns)}")
print("\nFirst 5 rows:")
metrics_df.head()
````

## 5. Basic Analytics

````python
from commit_relay import MetricsAggregator

aggregator = MetricsAggregator(client)

# Get comprehensive summary
summary = aggregator.get_comprehensive_summary(hours=24)

print("24-Hour Summary:")
print(f"\nWorker Stats:")
for key, value in summary['worker_stats'].items():
    print(f"  {key}: {value}")

print(f"\nTask Stats:")
for key, value in summary['task_stats'].items():
    print(f"  {key}: {value}")
````

## 6. Visualizations

````python
from commit_relay import DashboardVisualizer
import matplotlib.pyplot as plt

viz = DashboardVisualizer(client)

# Create comprehensive dashboard
viz.create_dashboard(hours=24)
plt.show()
````

## 7. Health Monitoring

````python
from commit_relay import HealthChecker

checker = HealthChecker(client)

# Get overall health
health = checker.get_overall_health()

print(f"Overall Health: {health['overall_status'].upper()}")
print(f"\nCheck Results:")

for check in health['checks']:
    status_symbol = {'healthy': '✓', 'warning': '!', 'critical': '✗', 'unknown': '?'}.get(check['status'], '?')
    print(f"  [{status_symbol}] {check['check']}: {check['message']}")
````

## 8. Trend Analysis

````python
from commit_relay import TrendAnalyzer

# Detect trend in success rate
if not metrics_df.empty and 'success_rate' in metrics_df.columns:
    trend = TrendAnalyzer.detect_trend(metrics_df['success_rate'])
    
    print("Success Rate Trend Analysis:")
    print(f"  Direction: {trend.get('direction', 'unknown').upper()}")
    print(f"  Slope: {trend.get('slope', 0):.4f}")
    print(f"  R-squared: {trend.get('r_squared', 0):.4f}")
    
    if trend.get('significant'):
        print(f"  ✓ Statistically significant (confidence: {trend.get('confidence', 0):.1%})")
    else:
        print(f"  - Not statistically significant")
````

## 9. Generate Reports

````python
from commit_relay import ReportGenerator

generator = ReportGenerator(client)

# Generate daily summary
report = generator.generate_daily_summary(hours=24)
print(report)

# Export to markdown
generator.export_to_markdown('/tmp/commit_relay_report.md', hours=24)
print("\n✓ Report saved to /tmp/commit_relay_report.md")
````

## 10. Data Export

````python
from commit_relay import DataExporter

exporter = DataExporter(client)

# Export everything to a directory
exporter.export_all_to_directory('/tmp/commit_relay_exports')
print("✓ All data exported to /tmp/commit_relay_exports")
````
